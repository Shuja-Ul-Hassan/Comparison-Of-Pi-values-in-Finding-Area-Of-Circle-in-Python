```python
import math

def areaofcircle1(radius):#defines the function area of a circle and takes radius as an argument

    P1=math.pi #approximately the value of pi

    formula=P1*radius**2 #googled the formula to find the area of circle
    
    area=round(formula,5) #rounded the answer of the formula to five decimal values for clarity

    return f"This is Your area: {area}" #returns area of circle
```

```python
value1=areaofcircle1(3)
print(value1)
area1=float(value1[-8:])
# area of a circle with radius (3)
```

```python
def areaofcircle2(radius):
    P1=22/7 
    formula=P1*radius**2
    area=round(formula,5)

    return f"This is Your area: {area}"
```

```python
value2=areaofcircle2(3)
print(value2)
area2=float(value2[-8:])
```

```python
import matplotlib.pyplot as plt
labels=["math.pi","22/7"]
values=[area1,area2]
plt.bar(labels,values,color=['blue','orange'])
plt.ylim(28.25, 28.30)

plt.title("Comparison of Pi Values")
plt.ylabel("Value")
plt.show()
```
